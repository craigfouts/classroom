```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = ['function 10'] + [f'function 0{i}' for i in range(9, 0, -1)]
# exp1 = np.array([4, 3, 1, 3.8, 2.8, 1.2, 1.4, 4.6, 2, 1])
exp2 = np.array([1, 3, 1, 4.4, 2.8, 1.2, 1.4, .2, 2, 1])
fig, ax = plt.subplots()
fig.set_size_inches(3, 4)
ax.barh(x, exp2)
ax.xaxis.tick_top()
ax.xaxis.set_label_position('top') 
ax.set_xlabel('Number of genes')
plt.show()
```
